```python
import pandas as pd

# Lire le fichier CSV
df = pd.read_csv(r'C:\Users\Clement\Downloads\019HexaSmal.csv', sep=';')

# Afficher les colonnes disponibles
print("Colonnes du fichier :", df.columns.tolist())

# Nettoyer les noms de colonnes (enlever # et espaces)
df.columns = df.columns.str.replace('#', '').str.strip()

print("Colonnes après nettoyage :", df.columns.tolist())

# Convertir Code_postal en texte, nettoyer les espaces et ajouter 0 devant si besoin (format 5 chiffres)
df['Code_postal'] = df['Code_postal'].astype(str).str.strip().str.zfill(5)

# Sélectionner uniquement les colonnes Code_commune_INSEE et Code_postal
df_selectionne = df[['Code_commune_INSEE', 'Code_postal']].copy()

# Supprimer les doublons uniquement sur Code_commune_INSEE (garde la première occurrence)
df_unique = df_selectionne.drop_duplicates(subset=['Code_commune_INSEE'], keep='first')

# Nettoyer aussi Code_commune_INSEE pour enlever les espaces
df_unique['Code_commune_INSEE'] = df_unique['Code_commune_INSEE'].astype(str).str.strip()

# Sauvegarder le résultat EN FORÇANT le type string pour éviter la conversion en int
df_unique.to_csv(r'C:\Users\Clement\OneDrive\Bureau\L3 info\Projet info\INSEE code.csv', 
                 sep=';', index=False, encoding='utf-8')

# Afficher les statistiques
print(f"\n=== RÉSULTATS ===")
print(f"Nombre total de lignes initiales : {len(df_selectionne)}")
print(f"Nombre de codes INSEE uniques : {len(df_unique)}")
print(f"Doublons supprimés : {len(df_selectionne) - len(df_unique)}")
print(f"\nAperçu des 10 premières lignes (format 5 chiffres) :")
print(df_unique.head(10))
print(f"\nType de Code_postal : {df_unique['Code_postal'].dtype}")
print(f"Exemples de codes postaux : {df_unique['Code_postal'].head(5).tolist()}")
print("\n✓ Fichier sauvegardé avec succès !")
```

```python
import pandas as pd

dvf = pd.read_csv(r'C:\Users\Clement\OneDrive\Bureau\L3 info\Projet info\dataVilleAnnee\dvf_consolidated_dataset.csv', sep=',')

insee_code = pd.read_csv(r'C:\Users\Clement\OneDrive\Bureau\L3 info\Projet info\INSEE code.csv', sep=';', dtype={'Code_postal': str})

insee_code['Code_commune_INSEE'] = insee_code['Code_commune_INSEE'].str.strip()
dvf['INSEE_COM'] = dvf['INSEE_COM'].astype(str).str.strip()

insee_code['Code_postal'] = insee_code['Code_postal'].str.strip()


dvf_avec_postal = dvf.merge(insee_code, left_on='INSEE_COM', right_on='Code_commune_INSEE', how='left')

dvf_avec_postal = dvf_avec_postal.drop(columns=['Code_commune_INSEE'])
dvf_avec_postal.to_csv(r'C:\Users\Clement\OneDrive\Bureau\L3 info\Projet info\dataVilleAnnee\dvf_avec_code_postal.csv', 
                       sep=',', index=False)
```
